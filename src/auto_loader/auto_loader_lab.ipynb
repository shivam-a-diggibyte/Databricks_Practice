```sql
CREATE CATALOG IF NOT EXISTS dev_catalog;
CREATE SCHEMA IF NOT EXISTS dev_catalog.bronze;
CREATE VOLUME IF NOT EXISTS dev_catalog.bronze.raw_volume;
```

```sql
%python
source_path = "/Volumes/dev_catalog/bronze/raw_volume/employees/"
schema_path = "/Volumes/dev_catalog/bronze/raw_volume/schema/"
checkpoint_path = "/Volumes/dev_catalog/bronze/raw_volume/checkpoint/"

df = (
    spark.readStream # creating stream input
        .format("cloudFiles") # use auto loader
        .option("cloudFiles.format", "json") # file is json
        .option("cloudFiles.schemaLocation", schema_path) # store schema location at this place
        .load(source_path) # load data from this location
)
```

```sql
%python
# write stream to a delta table

query = (
    df.writeStream
        .format("delta")
        .option("checkpointLocation", checkpoint_path)
        .option("mergeSchema", "true")
        .outputMode("append")
        .trigger(availableNow=True)
        .toTable("dev_catalog.bronze.employee_bronze")
)

query.awaitTermination()
```
